```python
# main_project.ipynb

from car_rental import CarRental, Customer

def main():
    shop = CarRental(10)
    customer = Customer()

    while True:
        print("""
        ===== CAR RENTAL MENU =====
        1. Display available cars
        2. Rent car(s) on hourly basis ($10/hour)
        3. Rent car(s) on daily basis ($40/day)
        4. Rent car(s) on weekly basis ($200/week)
        5. Return car(s)
        6. Exit
        """)
        
        choice = input("Enter your choice (1-6): ")

        if choice == '1':
            shop.display_stock()

        elif choice == '2':
            customer.rental_basis = 1
            customer.cars = customer.request_car()
            customer.rental_time = shop.rent_hourly(customer.cars)

        elif choice == '3':
            customer.rental_basis = 2
            customer.cars = customer.request_car()
            customer.rental_time = shop.rent_daily(customer.cars)

        elif choice == '4':
            customer.rental_basis = 3
            customer.cars = customer.request_car()
            customer.rental_time = shop.rent_weekly(customer.cars)

        elif choice == '5':
            bill = shop.return_car(customer.return_car())
            customer.rental_time, customer.rental_basis, customer.cars = None, 0, 0

        elif choice == '6':
            print("Thank you for using our service!")
            break
        else:
            print("[ERROR] Invalid choice. Please try again.")

```

```python
if __name__ == "__main__":
    main()
    
    
```
